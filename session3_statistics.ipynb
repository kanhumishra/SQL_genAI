```python
#Covariance 
l1 = list(range(1,11))
l2 = list(range(21,31))
arr1 = np.array(l1)
arr2 = np.array(l2)
arr1_mean= np.mean(arr1)
arr2_mean= np.mean(arr2)
arr11 = arr1-arr1_mean
arr22 = arr2-arr2_mean
np.sum(arr11*arr22)/len(arr1)
```
